```python
import pygame 
import sys
import random 
```

## Game Configuration 
- we are trying to configure the game this includes the widthe of the screen, the frame rate, the font size

```python
# set the configuration for the gaming window size ! 
SCREEN_WIDTH = 1200
SCREEN_HEIGHT = 900 



# to control the frame perseconds, how many updates in the for loop, but we need to configure the clock first 

clock = pygame.time.Clock()
FPS = 60

font_size = 72 

# configure the font to be the defualt and set it to 72 which is the size ! 
font = pygame.font.Font(None,font_size) 
```

## Cart Configuration 
- this focus on the configuration of the cart itself, the dimensions of the cart and the position.


```python
CART_Y_OFFSET = 50
CART_WIDTH = 150
CART_HEIGHT = 50

# To draw the rectangel you need the top left corner of it, which can be described using x and y.
## y is measured from the top of the screen, so we subtract from the hight the offset and the cart height to get the y of the top left
### of this rectangle.
cart_y = SCREEN_HEIGHT- CART_Y_OFFSET - CART_HEIGHT 
# x of the cart, in case of middle we will have to subtract the cart width too ? why ! imagine a cart that have the same size of the 
## window this will result cart out of the window itself.
cart_x = SCREEN_WIDTH // 2 - CART_WIDTH // 2

# this is not a real speed, it is just a multiplier that will be used to move the box, in case you click right or left
## the cart will use this factor to get a bouns ! 
CART_SPEED  =  5
```

```python
# ObjectA properties
OBJECT_SIZE = 15
objectA_x = random.randint(0, SCREEN_WIDTH - OBJECT_SIZE)
objectA_y = cart_y  # keep it above cart area
```

```python
# render the screen with the configuration we choose ~ 
screen = pygame.display.set_mode((SCREEN_WIDTH, SCREEN_HEIGHT))

# set the score initially to zero
score = 0

# --- NEW: timer for red dot ---
OBJECT_TIME_LIMIT = 1500  # 0.5 seconds
last_spawn_time = pygame.time.get_ticks()

while True:
    
    clock.tick(FPS)

    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            pygame.quit()
            sys.exit()

    keys = pygame.key.get_pressed()
    if keys[pygame.K_LEFT]:
        cart_x -= CART_SPEED
    if keys[pygame.K_RIGHT]:
        cart_x += CART_SPEED

    # keep object A inside the window 
    if objectA_x < 0:
        objectA_x = 0
    if objectA_x + OBJECT_SIZE > SCREEN_WIDTH:
        objectA_x = SCREEN_WIDTH - OBJECT_SIZE

    # Keep cart inside the window
    if cart_x < 0:
        cart_x = 0
    if cart_x + CART_WIDTH > SCREEN_WIDTH:
        cart_x = SCREEN_WIDTH - CART_WIDTH

    screen.fill((30, 30, 30))

    # Draw the cart
    pygame.draw.rect(screen, (0, 200, 255), (cart_x, cart_y, CART_WIDTH, CART_HEIGHT))

    # Collision check
    cart_rect = pygame.Rect(cart_x, cart_y, CART_WIDTH, CART_HEIGHT)
    objectA_rect = pygame.Rect(objectA_x, objectA_y, OBJECT_SIZE, OBJECT_SIZE)

    if cart_rect.colliderect(objectA_rect):
        score += 1
        objectA_x = random.randint(0, SCREEN_WIDTH - OBJECT_SIZE)
        last_spawn_time = pygame.time.get_ticks()   # reset timer

    # --- NEW: check if time expired ---
    current_time = pygame.time.get_ticks()
    if current_time - last_spawn_time > OBJECT_TIME_LIMIT:
        objectA_x = random.randint(0, SCREEN_WIDTH - OBJECT_SIZE)
        last_spawn_time = current_time

    # Draw score
    score_surface = font.render(f"Score: {score}", True, (255, 255, 255))
    screen.blit(score_surface, (20, 20))

    # Draw object A
    pygame.draw.rect(screen, (255, 50, 50), (objectA_x, objectA_y, OBJECT_SIZE, OBJECT_SIZE))

    pygame.display.flip()
```

```python
import pygame 
import sys
import random 

# ----------------------- INITIALIZATION -----------------------
pygame.init()
pygame.font.init()

# Game Configuration
SCREEN_WIDTH = 1200
SCREEN_HEIGHT = 900

clock = pygame.time.Clock()
FPS = 60

font_size = 72
font = pygame.font.Font(None, font_size)

# Cart Configuration
CART_Y_OFFSET = 50
CART_WIDTH = 150
CART_HEIGHT = 50

cart_y = SCREEN_HEIGHT - CART_Y_OFFSET - CART_HEIGHT
cart_x = SCREEN_WIDTH // 2 - CART_WIDTH // 2
CART_SPEED = 10

# Objects
OBJECT_SIZE = 15

# Red object (now BAD → -points)
objectA_x = random.randint(0, SCREEN_WIDTH - OBJECT_SIZE)
objectA_y = cart_y

# Green object (now GOOD → +points)
objectB_x = random.randint(0, SCREEN_WIDTH - OBJECT_SIZE)
objectB_y = cart_y

# Timer for both objects
OBJECT_TIME_LIMIT = 1000  # 1 second
last_spawn_time = pygame.time.get_ticks()

# Score
score = 0

# Render screen
screen = pygame.display.set_mode((SCREEN_WIDTH, SCREEN_HEIGHT))


# ----------------------- GAME LOOP -----------------------
while True:
    
    clock.tick(FPS)

    # Event handling
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            pygame.quit()
            sys.exit()

    # Move cart
    keys = pygame.key.get_pressed()
    if keys[pygame.K_LEFT]:
        cart_x -= CART_SPEED
    if keys[pygame.K_RIGHT]:
        cart_x += CART_SPEED

    # Keep cart inside screen
    if cart_x < 0:
        cart_x = 0
    if cart_x + CART_WIDTH > SCREEN_WIDTH:
        cart_x = SCREEN_WIDTH - CART_WIDTH

    # Keep objects inside screen
    objectA_x = max(0, min(objectA_x, SCREEN_WIDTH - OBJECT_SIZE))
    objectB_x = max(0, min(objectB_x, SCREEN_WIDTH - OBJECT_SIZE))

    # Background
    screen.fill((30, 30, 30))

    # Draw cart
    pygame.draw.rect(screen, (0, 200, 255), (cart_x, cart_y, CART_WIDTH, CART_HEIGHT))

    # Create rects for collision
    cart_rect = pygame.Rect(cart_x, cart_y, CART_WIDTH, CART_HEIGHT)
    objectA_rect = pygame.Rect(objectA_x, objectA_y, OBJECT_SIZE, OBJECT_SIZE)
    objectB_rect = pygame.Rect(objectB_x, objectB_y, OBJECT_SIZE, OBJECT_SIZE)

    # ---- COLLISION WITH RED (BAD → -1) ----
    if cart_rect.colliderect(objectA_rect):
        score -= 1
        objectA_x = random.randint(0, SCREEN_WIDTH - OBJECT_SIZE)
        last_spawn_time = pygame.time.get_ticks()

    # ---- COLLISION WITH GREEN (GOOD → +1) ----
    if cart_rect.colliderect(objectB_rect):
        score += 1
        objectB_x = random.randint(0, SCREEN_WIDTH - OBJECT_SIZE)
        last_spawn_time = pygame.time.get_ticks()

    # ---- TIMEOUT RESPAWN ----
    current_time = pygame.time.get_ticks()

    if current_time - last_spawn_time > OBJECT_TIME_LIMIT:
        objectA_x = random.randint(0, SCREEN_WIDTH - OBJECT_SIZE)
        objectB_x = random.randint(0, SCREEN_WIDTH - OBJECT_SIZE)
        last_spawn_time = current_time

    # Draw score
    score_surface = font.render(f"Score: {score}", True, (255, 255, 255))
    screen.blit(score_surface, (20, 20))

    # Draw red object (bad)
    pygame.draw.rect(screen, (255, 50, 50), (objectA_x, objectA_y, OBJECT_SIZE, OBJECT_SIZE))

    # Draw green object (good)
    pygame.draw.rect(screen, (50, 255, 50), (objectB_x, objectB_y, OBJECT_SIZE, OBJECT_SIZE))

    pygame.display.flip()
```
