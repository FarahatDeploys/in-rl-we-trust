## Import the pygame module and system module 

pygame will be responsible for triggering graphics, sounds, and receive inputs from the keyboard to do this it will rely on 




```python

import pygame 
import sys 
```

Pygame does the following:

- Converts Python objects into C structures

- Calls SDL (Simple DirectMedia Layer — C library)

- Manages Pygame-specific features (surfaces, fonts, events)


Pygame provides a high-level API that wraps SDL, which communicates with low-level OS subsystems and hardware drivers.



However before using it we must do some initizliation to : 

- Allocates memory for surfaces, audio, and fonts.

- Opens device handles (graphics device, audio device, joysticks) / these pointers are normally assigned to void or null.

- Registers callbacks with the OS (so events can be captured).

- Starts internal threads/timers in SDL for rendering and audio mixing.

```python
pygame.init()
```

Now the pygame modules is ready to be used from high level

first we set the window size which is called window mode 


```python
screen = pygame.display.set_mode((800,600))
```

To make this screen, object we can close, or do any process with we need a while loop that keeps checking 
all the events by calling pygame.event.get(), if a quit event get called, the pygame will exit and the system too ! 
otherwise we fill screen with the black bacground

```python
while True:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            pygame.quit()
            sys.exit()
    screen.fill((0,0,0))
    pygame.display.flip()
```

now let's build a more complex game, this one should act as a basic counter ! 

```python
screen = pygame.display.set_mode((1200,900))
clock = pygame.time.Clock()
counter = 0
pygame.font.init()  
font = pygame.font.SysFont(None, 72) 

while True:
    clock.tick(30)
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            pygame.quit()
            sys.exit()

    keys = pygame.key.get_pressed()

    if keys[pygame.K_LEFT]:
        counter = counter - 1 
    if keys[pygame.K_RIGHT]:
        counter = counter + 1

    screen.fill((0, 0, 0))
    counter_text = font.render(str(counter), True, (255, 255, 255))  # white color
    screen.blit(counter_text, (100, 100))  # draw at (x=100, y=100)

    pygame.display.flip()
```
